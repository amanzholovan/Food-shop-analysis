```python
import folium

m = folium.Map(location=[42.9, 71.37], zoom_start=13)

for idx, row in df_geo.iterrows():
    folium.CircleMarker(
        location=[row['Широта'], row['Долгота']],
        radius=5,
        popup=row['Наименование'],
        color='blue',
        fill=True
    ).add_to(m)

m
```

```python
m2 = folium.Map(location=[42.9, 71.37], zoom_start=13)

for idx, row in df_geo.iterrows():
    rating = row['Рейтинг']
    if pd.isna(rating):
        color = 'gray'
    elif rating >= 4.5:
        color = 'green'
    elif rating >= 4.0:
        color = 'orange'
    else:
        color = 'red'
    
    folium.CircleMarker(
        location=[row['Широта'], row['Долгота']],
        radius=6,
        popup=f"{row['Наименование']}: {rating}",
        color=color,
        fill=True,
        fill_color=color
    ).add_to(m2)

m2
```

## Conclusions

Wanted to look at the semi-finished foods market in Taraz instead of using
another Kaggle dataset, so I scraped it myself from 2GIS.

Got 53 listings for "магазин полуфабрикаты" shops, but a few (seafood shops mostly)
got pulled in by mistake because of overlapping category tags. Filtered down
to 44 real ones.

Only 28 out of 44 have any rating. Average is 4.4, which is pretty good, but
most shops have barely any reviews - median is 3. 

Mapped everything out and there's a clear gap in the northwest and southeast
parts of the city - way fewer shops there than in the center.

Next time I'd want real review text, some way to check if low visibility
actually means low foot traffic, and competitor pricing to see if there's
a gap there too.
